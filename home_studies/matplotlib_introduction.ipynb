# Matplotlib Intorduction 

```python
import matplotlib.pyplot as plt
import numpy as np
```

## Drawing a line in matplotlib

`linspace()` is a numpy function that generates a sequence of evenly spaced numbers between two endpoints.

```python
x = np.linspace(0, 1, 5)
print(x)
```

calculating y using the line equation `x + y = 1`

```python
y = 1 - x
```

using `plt` to plot the line.

```python
plt.plot(x, y, label="x + y = 1")
plt.xlabel('x')
plt.ylabel("y")

plt.title("Straight Line")
plt.legend()
plt.grid(True)
```

## Drawing multiple lines in matplotlib

```python
x = np.linspace(-1, 1, 100)
y1 = 1- x
y2 = x
```

```python
plt.plot(x, y1, label='x + y = 1', color='blue')
plt.plot(x, y2, label='x - y = 0', color='red')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Multiple Lines')
plt.legend()
plt.grid(True)
plt.axis('equal')
```

## Using `matplotlib` in Deep Learing

Define the neuron parameters that correspond to the line x + y - 1 = 0

```python
w1, w2, b = 1, 1, -1

limitPoint = 6

x = np.linspace(-limitPoint, limitPoint, 100)
y = np.linspace(-limitPoint, limitPoint, 100)
X, Y = np.meshgrid(x, y)

print(X)
print(Y)
print(X.shape)
print(Y.shape)
```

```python
Z = w1 * X + w2 * Y + b
output = (Z <= 0).astype(int)
```

```python
plt.contour(X, Y, Z, levels=[0], colors=["blue"], linewidths=2)

plt.contourf(X, Y, output, levels=[0.5, 1.5], colors=["yellow"], alpha=0.5)

plt.title('Neuron decision region: output = 1 highlighted')
plt.grid(True)
plt.xlim(-limitPoint, limitPoint)
plt.ylim(-limitPoint, limitPoint)
plt.xlabel('x')
plt.ylabel('y')
plt.axis('equal')
plt.legend()

ax = plt.gca()
# Move left and bottom spines (axes lines) to the center
ax.spines['left'].set_position('center')
ax.spines['bottom'].set_position('center')
```
